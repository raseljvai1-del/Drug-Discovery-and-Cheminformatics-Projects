### Different between Descriptor vs Fingerprint

> **Descriptor = "What properties does this molecule have?"**
>
> **Fingerprint = "What structural patterns does this molecule contain?"**
>
>
> **Molecular descriptors quantify the characteristics of a molecule, while molecular fingerprints encode its structural patterns into a machine-readable vector for tasks such as similarity comparison and searching.**


```text
                  MOLECULE
                     │
          ┌──────────┴──────────┐
          ↓                     ↓
     DESCRIPTORS            FINGERPRINTS
          │                     │
          ↓                     ↓
   "What is it like?"    "What patterns does
                          it contain?"
          │                     │
          ↓                     ↓
    MW, LogP, TPSA       Morgan, MACCS,
    HBD, HBA, etc.       RDKit, etc.
          │                     │
          ↓                     ↓
    Property analysis      Similarity search
```


### Descriptor vs Fingerprint

|                           | **Molecular Descriptor**                             | **Molecular Fingerprint**                               |
| ------------------------- | ---------------------------------------------------- | ------------------------------------------------------- |
| **Input**                 | Molecule                                             | Molecule                                                |
| **Output**                | Numerical values                                     | Vector of bits or counts                                |
| **Main idea**             | Describes **properties** of a molecule               | Describes **structural patterns** of a molecule         |
| **What does it tell us?** | "How big, polar, hydrophobic, flexible, etc. is it?" | "What atoms, bonds, and substructures does it contain?" |
| **Example**               | MW = 180.16                                          | `[0, 1, 0, 1, 0, ...]`                                  |
| **Examples**              | MW, LogP, TPSA, HBD, HBA                             | Morgan, MACCS, RDKit, Atom Pair                         |
| **Human-readable?**       | Usually **yes**                                      | Usually **no**                                          |
| **Number of values**      | Usually a relatively small set                       | Often hundreds/thousands of values                      |
| **Represents**            | Molecular characteristics                            | Molecular structure/patterns                            |
| **Common use**            | Property analysis, ML features                       | Similarity, search, clustering                          |
| **Example question**      | "How polar is this molecule?"                        | "How structurally similar are these two molecules?"     |

---

#### 1. Molecular Descriptor = Molecular "Measurements"
>
> **Descriptor → numerical measurement of a molecular property/characteristic.**
>
>
#### 2. Fingerprint = Molecular "Pattern ID"
>
> "Which structural patterns are present in this molecule?"
>
> **Fingerprint → numerical representation of molecular structure/patterns.**
